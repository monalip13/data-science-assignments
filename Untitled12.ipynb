```python
# lab13_california_regression.py
# Lab 13: Exploring Data and Insights
# Purpose: load a public dataset suitable for multiple linear regression,
# build and evaluate a multiple linear regression model, produce diagnostics,
# and save outputs (figures and csv) for inclusion in a Word report.
#
# Required packages:
# pip install pandas numpy scikit-learn statsmodels matplotlib seaborn scipy

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from statsmodels.stats.outliers_influence import variance_inflation_factor
from scipy import stats

# ---------------------------
# 1. Use an AI tool to help you locate and select a public dataset appropriate 
# for multiple linear regression (it should include at least two predictor variables and one numeric target).
# load the dataset.
# ---------------------------

# Fetch the California Housing dataset (public, appropriate for regression).
# It contains a numeric target (MedianHouseValue) and multiple numeric predictors.
cal = fetch_california_housing(as_frame=True)  # returns a Bunch with .frame
df = cal.frame.copy()  # DataFrame with features + target

# Rename target column to a clearer name
df = df.rename(columns={"MedHouseVal": "MedHouseValue"})  # per sklearn naming

# Show the first 10 rows in console and save a CSV snapshot for screenshotting in Word.
head10 = df.head(10)
print("First 10 rows of the dataset:")
print(head10)
# Save the head as CSV so you can include a screenshot or embed table in the Word document.
os.makedirs("outputs", exist_ok=True)
head10.to_csv("outputs/data_head10.csv", index=False)

# ---------------------------
# 2. Document dataset basics
# ---------------------------

print("\nDataset shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nTarget summary statistics:")
print(df["MedHouseValue"].describe())

# ---------------------------
# 3. Research question (defined with AI) and variables selection
# ---------------------------
# Research question used for this analysis:
# "How do median income, average rooms, and housing age (together with population and average occupancy)
# influence median house value in California neighborhoods?"
#
# Predictors chosen (minimum two as required): MedInc, AveRooms, HouseAge.
# Additional predictors included for robustness / interpretation: Population, AveOccup.

predictor_cols = ["MedInc", "AveRooms", "HouseAge", "Population", "AveOccup"]
target_col = "MedHouseValue"

# ---------------------------
# 4. Data types, missing values, outliers
# ---------------------------

# Check missing values
print("\nMissing values per column:")
print(df[predictor_cols + [target_col]].isna().sum())

# There are no missing values in sklearn's California housing dataset by default.
# But we'll still defensively handle any unexpected missing values.
df = df[predictor_cols + [target_col]].copy()
df = df.dropna()  # drop rows with missing values if any (simple approach documented in report)
print("\nAfter dropping NAs, shape:", df.shape)

# Outlier handling: use z-score trimming for extreme outliers on numeric columns (conservative).
# We'll remove rows where any predictor or target has |z| > 4 (very extreme).
z_scores = np.abs(stats.zscore(df))
# For small numerical instability, handle constant columns
z_scores = np.nan_to_num(z_scores)
outlier_mask = (z_scores > 4).any(axis=1)
print("\nNumber of extreme-outlier rows to be removed (|z|>4):", outlier_mask.sum())
df = df.loc[~outlier_mask].reset_index(drop=True)
print("After outlier removal, shape:", df.shape)

# ---------------------------
# 5. Train-test split
# ---------------------------
# Split data so we can evaluate on unseen data (70% train, 30% test)
X = df[predictor_cols]
y = df[target_col]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)

# ---------------------------
# 6. Scaling predictors (for coefficient comparison)
# ---------------------------
# We'll keep a version of unscaled data for 'raw-coefficient' interpretation,
# and build a standardized model to aid interpretation of relative variable importance.
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=predictor_cols, index=X_train.index)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=predictor_cols, index=X_test.index)

# ---------------------------
# 7. Build the multiple linear regression model (statsmodels)
# ---------------------------

# Add constant term for intercept
X_train_sm = sm.add_constant(X_train)                 # raw model (unstandardized)
X_train_sm_scaled = sm.add_constant(X_train_scaled)   # standardized model

# Fit OLS on unstandardized predictors (keeps original units)
model_raw = sm.OLS(y_train, X_train_sm).fit()

# Fit OLS on standardized predictors (coeffs are comparable)
model_std = sm.OLS(y_train, X_train_sm_scaled).fit()

# Save model summaries to text files
with open("outputs/model_raw_summary.txt", "w") as f:
    f.write(model_raw.summary().as_text())
with open("outputs/model_std_summary.txt", "w") as f:
    f.write(model_std.summary().as_text())

print("\nRaw model summary (first lines):")
print(model_raw.summary().as_text()[:800])  # print first 800 chars

# ---------------------------
# 8. Define hypotheses (explicit)
# ---------------------------
# Null hypothesis (H0): None of the included predictors (MedInc, AveRooms, HouseAge, Population, AveOccup)
#                        have a linear relationship with the median house value (i.e., their coefficients = 0).
# Alternative (Ha): At least one predictor has a coefficient not equal to zero (predicts MedHouseValue).
#
# We will discuss single-coefficient hypotheses (e.g., H0: beta_MedInc = 0 vs Ha: beta_MedInc != 0)
# using each predictor's t-test p-value from the model summary.

# ---------------------------
# 9. Multicollinearity check: VIF
# ---------------------------

# Compute VIF on scaled predictors for interpretability
X_vif = sm.add_constant(X_train_scaled)
vif_data = pd.DataFrame({
    "feature": X_vif.columns,
    "VIF": [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
})
print("\nVIF results:")
print(vif_data)
vif_data.to_csv("outputs/vif.csv", index=False)

# ---------------------------
# 10. Model evaluation on test set
# ---------------------------

# Prepare test data for prediction (use unscaled model predictions for RMSE in original units)
X_test_sm = sm.add_constant(X_test)
y_pred_raw = model_raw.predict(X_test_sm)

# Calculate test metrics: R^2 (on test), RMSE
from sklearn.metrics import mean_squared_error, r2_score
rmse_test = mean_squared_error(y_test, y_pred_raw, squared=False)
r2_test = r2_score(y_test, y_pred_raw)
print(f"\nTest RMSE: {rmse_test:.4f}")
print(f"Test R^2: {r2_test:.4f}")

# ---------------------------
# 11. Visualizations (saved to outputs/)
# ---------------------------

sns.set(style="whitegrid", font_scale=1.1)

# 11a. Pairplot (scatter matrix) of predictors + target (sample for speed)
sample_df = df.sample(n=min(len(df), 2000), random_state=42)  # limit size for pairplot
pairplot_vars = predictor_cols + [target_col]
pp = sns.pairplot(sample_df[pairplot_vars], corner=True, kind="reg", plot_kws={"line_kws": {"color": "red"}, "scatter_kws": {"s": 8}})
pp.fig.suptitle("Pairplot: predictors vs target (sample)", y=1.02)
pp.savefig("outputs/pairplot_sample.png", dpi=150)
plt.close()

# 11b. Coefficient plot (standardized model)
coef = model_std.params.drop("const")
coef_ci = model_std.conf_int().loc[coef.index]
coef_df = pd.DataFrame({
    "coef": coef.values,
    "ci_low": coef_ci[0].values,
    "ci_high": coef_ci[1].values
}, index=coef.index)
coef_df = coef_df.sort_values("coef")

plt.figure(figsize=(8, 4))
plt.errorbar(coef_df["coef"], coef_df.index, xerr=[coef_df["coef"] - coef_df["ci_low"], coef_df["ci_high"] - coef_df["coef"]], fmt='o')
plt.axvline(0, color='gray', linewidth=0.8)
plt.title("Coefficient plot (standardized predictors) — OLS")
plt.xlabel("Standardized coefficient (units of std dev)")
plt.tight_layout()
plt.savefig("outputs/coef_plot_std.png", dpi=150)
plt.close()

# 11c. Residuals plot (train)
residuals = model_raw.resid
fitted = model_raw.fittedvalues

plt.figure(figsize=(8,5))
plt.scatter(fitted, residuals, s=10)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel("Fitted values (train)")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted (train)")
plt.tight_layout()
plt.savefig("outputs/residuals_vs_fitted_train.png", dpi=150)
plt.close()

# 11d. QQ-plot of residuals
fig = sm.graphics.qqplot(residuals, line='45', fit=True)
fig.suptitle("QQ-plot of residuals (train)")
fig.savefig("outputs/qqplot_residuals.png", dpi=150)
plt.close()

# 11e. Observed vs Predicted (test)
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred_raw, s=10)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel("Observed MedHouseValue (test)")
plt.ylabel("Predicted MedHouseValue (test)")
plt.title("Observed vs Predicted (test set)")
plt.tight_layout()
plt.savefig("outputs/observed_vs_predicted_test.png", dpi=150)
plt.close()

# ---------------------------
# 12. Save important outputs
# ---------------------------

# Save model coefficients table
coef_table = pd.concat([model_raw.params, model_raw.bse, model_raw.pvalues], axis=1)
coef_table.columns = ["coef", "std_err", "p_value"]
coef_table.to_csv("outputs/model_raw_coefs.csv")

# Save standardized coefficients
coef_std_table = pd.concat([model_std.params, model_std.bse, model_std.pvalues], axis=1)
coef_std_table.columns = ["coef_std", "std_err_std", "p_value_std"]
coef_std_table.to_csv("outputs/model_std_coefs.csv")

# Save test predictions
pred_df = pd.DataFrame({"y_test": y_test, "y_pred": y_pred_raw})
pred_df.to_csv("outputs/test_predictions.csv", index=False)

# Save a brief JSON-like summary
summary = {
    "train_observations": int(len(X_train)),
    "test_observations": int(len(X_test)),
    "model_raw_r2_train": float(model_raw.rsquared),
    "model_raw_adj_r2_train": float(model_raw.rsquared_adj),
    "test_r2": float(r2_test),
    "test_rmse": float(rmse_test)
}
import json
with open("outputs/summary.json", "w") as f:
    json.dump(summary, f, indent=2)

print("\nAll outputs saved to the 'outputs' folder (figures, csv, model summaries).")
print("Files to include in your Word document: outputs/data_head10.csv (or screenshot), outputs/model_raw_summary.txt, outputs/model_std_summary.txt, and all images in outputs/*.png")
```

```python
# Question 5: Build a multiple linear regression model using at least two predictors.
# ==========================================
# ==========================================
# Import necessary library
# ==========================================
import statsmodels.api as sm

# ==========================================
# Define predictors (X) and target (y)
# ==========================================
# Using actual column names from your dataset
X = df[['median_income', 'total_rooms', 'population']]  # Predictor variables
y = df['median_house_value']                            # Target variable

# ==========================================
# Add a constant term to include intercept in the model
# ==========================================
X = sm.add_constant(X)

# ==========================================
# Build and fit the multiple linear regression model
# ==========================================
model = sm.OLS(y, X).fit()

# ==========================================
# Display regression summary to interpret coefficients and model performance
# ==========================================
print(model.summary())
```

```python
# Question 6: Clearly define and state your null and alternative hypotheses based on your model.
# Hypotheses establish the relationship we are testing between predictors (median_income, total_rooms, and population) and the target (median_house_value).

# Null Hypothesis (H₀):
# There is no statistically significant linear relationship between the predictors (median_income, total_rooms, and population) 
# and the dependent variable (median_house_value) in California.

# Alternative Hypothesis (H₁):
# At least one of the predictors (median_income, total_rooms, or population) has a significant linear effect on median_house_value.
```
