```python
# --- Load standard libraries and dataset ---
from sklearn import datasets            # load the Iris dataset (built-in)
import pandas as pd                    # for DataFrame handling
import numpy as np                     # numeric utilities
import matplotlib.pyplot as plt        # plotting
from statsmodels.formula.api import ols# OLS model interface for ANOVA
import statsmodels.api as sm           # statsmodels for ANOVA table
from statsmodels.stats.multicomp import pairwise_tukeyhsd  # Tukey HSD
from scipy import stats                # Shapiro, Levene, f_oneway - common tests

# --- Load Iris and create DataFrame ---
iris = datasets.load_iris(as_frame=True)   # load iris into a frame (pandas included)
df = iris.frame.copy()                     # copy so we can modify safely
# create a categorical 'species' column from target numeric and target names
df['species'] = pd.Categorical(iris.target_names[iris.target])

# rename columns to simpler names for clarity (optional but helpful)
df = df.rename(columns={
    'sepal length (cm)': 'sepal_length',
    'sepal width (cm)': 'sepal_width',
    'petal length (cm)': 'petal_length',
    'petal width (cm)': 'petal_width'
})

# --- Preview data (display first 10 rows) ---
print("First 10 rows of the dataset:")
print(df.head(10))   # show first 10 rows for dataset preview (screenshot this for report)

# --- Prepare groups for later tests: dictionary keyed by species ---
groups = {species: df.loc[df['species'] == species, 'sepal_length'].values
          for species in df['species'].unique()}

# --- Visual checks: histograms per group (normality) ---
for species, values in groups.items():
    # make a histogram of sepal length for each species
    plt.figure(figsize=(6,4))
    plt.hist(values, bins=8)
    plt.title(f'Histogram of Sepal Length — {species}')
    plt.xlabel('Sepal length (cm)')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

# --- Boxplot: sepal length by species for spread and medians ---
plt.figure(figsize=(7,5))
df.boxplot(column='sepal_length', by='species')   # grouped boxplot
plt.title('Boxplot of Sepal Length by Species')
plt.suptitle('')  # remove default subtitle added by boxplot
plt.xlabel('Species')
plt.ylabel('Sepal length (cm)')
plt.tight_layout()
plt.show()

# --- Group means with standard deviation error bars (visual of central tendency) ---
means = df.groupby('species')['sepal_length'].mean()   # compute means
stds = df.groupby('species')['sepal_length'].std()     # compute standard deviations
plt.figure(figsize=(6,4))
x = np.arange(len(means))
plt.bar(x, means)
plt.errorbar(x, means, yerr=stds, fmt='none', capsize=5)  # sd as error bars
plt.xticks(x, means.index)
plt.title('Group means (Sepal length) with SD error bars')
plt.ylabel('Mean Sepal length (cm)')
plt.xlabel('Species')
plt.tight_layout()
plt.show()

# --- Assumption checks: Shapiro-Wilk for normality (per group) ---
normality_results = {}
for species, values in groups.items():
    stat, p = stats.shapiro(values)                 # Shapiro-Wilk test
    normality_results[species] = {'W': stat, 'p-value': p}

# print normality test results
print("Shapiro-Wilk normality test per species (W, p-value):")
for s, res in normality_results.items():
    print(f"  {s}: W={res['W']:.4f}, p={res['p-value']:.4f}")

# --- Test homogeneity of variance: Levene's test ---
levene_stat, levene_p = stats.levene(*[v for v in groups.values()], center='median')
print("\nLevene test for homogeneity of variances:")
print(f"  statistic={levene_stat:.4f}, p-value={levene_p:.4f}")

# --- One-way ANOVA using statsmodels (recommended for table) ---
model = ols('sepal_length ~ C(species)', data=df).fit()  # fit OLS model with categorical species
anova_table = sm.stats.anova_lm(model, typ=2)           # type II ANOVA table
print("\nANOVA (statsmodels) table:")
print(anova_table)

# --- One-way ANOVA with scipy for F and p (double-check) ---
f_stat, f_p = stats.f_oneway(*[v for v in groups.values()])
print("\nANOVA (scipy.stats.f_oneway):")
print(f"  F-statistic={f_stat:.4f}, p-value={f_p:.6f}")

# --- Post-hoc testing: Tukey's HSD for pairwise comparisons ---
tukey = pairwise_tukeyhsd(endog=df['sepal_length'], groups=df['species'], alpha=0.05)
print("\nTukey HSD results:")
print(tukey.summary())   # printable table of pairwise comparisons

# Optionally save Tukey summary (makes it easy to attach to report)
# tukey_df = pd.DataFrame(data=tukey._results_table.data[1:], columns=tukey._results_table.data[0])
# tukey_df.to_csv('/mnt/data/tukey_summary.csv', index=False)
```
