```python
# =====================================================
# Robust Logistic Regression Pipeline for Marketing CSV
# =====================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_curve, auc
import matplotlib.pyplot as plt
import joblib
from pathlib import Path

# ---------------------------
# 1. Load CSV robustly
# ---------------------------
file_path = "/Users/patrickpatel/Downloads/Lab 12 - Marketing.csv"

# Try reading with comma first
try:
    df = pd.read_csv(file_path)
except Exception:
    # fallback: read entire line and split manually
    df = pd.read_csv(file_path, header=None)
    df = df[0].str.split(",", expand=True)

# Strip whitespaces in columns
df.columns = df.columns.str.strip()
print("Columns loaded:", df.columns.tolist())

# Check first few rows
print(df.head(3))

# ---------------------------
# 2. Detect target column
# ---------------------------
# Your CSV target is 'y' with 'yes'/'no'
target_col = None
for col in df.columns:
    if df[col].dropna().isin(['yes', 'no']).any():
        target_col = col
        break

if target_col is None:
    raise ValueError("Cannot detect target column with 'yes'/'no' values!")

print("Detected target column:", target_col)

# Encode target
df['y_bin'] = df[target_col].map({'yes':1, 'no':0})
df = df.dropna(subset=['y_bin'])
y = df['y_bin'].astype(int)

# ---------------------------
# 3. Detect numeric and categorical columns
# ---------------------------
numeric_cols = []
categorical_cols = []

for col in df.columns:
    if col in [target_col, 'y_bin']:
        continue
    try:
        pd.to_numeric(df[col].dropna())
        numeric_cols.append(col)
    except:
        categorical_cols.append(col)

print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)

# Convert numeric columns
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Drop rows with missing numeric values
df = df.dropna(subset=numeric_cols)

# ---------------------------
# 4. Prepare feature matrix
# ---------------------------
X = pd.get_dummies(df[numeric_cols + categorical_cols], drop_first=True)
X = X.fillna(0)
print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)

if X.shape[0] == 0 or y.shape[0] == 0:
    raise ValueError("No valid samples after preprocessing! Check CSV content.")

# ---------------------------
# 5. Train-test split
# ---------------------------
RANDOM_STATE = 42
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=RANDOM_STATE
)
print("Train/Test shapes:", X_train.shape, X_test.shape, y_train.shape, y_test.shape)

# ---------------------------
# 6. Scale numeric features
# ---------------------------
scaler = StandardScaler()
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])

# ---------------------------
# 7. Train Logistic Regression
# ---------------------------
model = LogisticRegression(max_iter=1000, class_weight='balanced', solver='liblinear', random_state=RANDOM_STATE)
model.fit(X_train, y_train)

# ---------------------------
# 8. Evaluate
# ---------------------------
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:,1]

acc = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred, digits=4)
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)

print(f"Accuracy: {acc:.4f}, AUC: {roc_auc:.4f}")
print("Confusion Matrix:\n", cm)
print("Classification Report:\n", report)

# ---------------------------
# 9. Save model, scaler, and metrics
# ---------------------------
OUTDIR = Path("output")
OUTDIR.mkdir(exist_ok=True)

joblib.dump(model, OUTDIR / "logistic_model.joblib")
joblib.dump(scaler, OUTDIR / "scaler.joblib")

with open(OUTDIR / "metrics.txt", "w") as f:
    f.write(f"Accuracy: {acc:.4f}\nAUC: {roc_auc:.4f}\n\n")
    f.write("Confusion Matrix:\n" + str(cm) + "\n\n")
    f.write("Classification Report:\n" + report + "\n")

print("All outputs saved in 'output' folder ✅")

# ---------------------------
# 10. Plot ROC curve
# ---------------------------
plt.figure(figsize=(7,6))
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {roc_auc:.3f})")
plt.plot([0,1],[0,1], linestyle='--', color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Logistic Regression")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(OUTDIR / "roc_curve.png", dpi=300)
plt.close()
```
