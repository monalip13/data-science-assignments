```python
#1. Load the dataset 'Chipotle.csv' using Pandas.
# Import libraries and load dataset
import pandas as pd
df = pd.read_excel('Chipotle.xls')
# Display first 5 rows to confirm load
print("Dataset Preview:")
print(df.head())
```

```python

#2. Convert the 'item_price' column to a numeric data type to enable analysis. 
# The column contains a "$" sign, so we strip it and convert to float
df['item_price'] = df['item_price'].replace('[\$,]', '', regex=True).astype(float)
print("\nConverted 'item_price' column (first 5 rows):")
print(df['item_price'].head())

```

```python
#3. Calculate and display the average price of items that contain 'chicken' in their name. 
# Filter rows containing 'chicken' (case-insensitive)
chicken_items = df[df['item_name'].str.contains('chicken', case=False, na=False)]

# Calculate average price
avg_chicken_price = chicken_items['item_price'].mean()

print(f"\nAverage price of chicken items: ${avg_chicken_price:.2f}")

```

```python
#4. Calculate and display the average price of items that contain 'steak' in their name. 
# Filter rows containing 'steak' (case-insensitive)
steak_items = df[df['item_name'].str.contains('steak', case=False, na=False)]

# Calculate average price
avg_steak_price = steak_items['item_price'].mean()

print(f"Average price of steak items: ${avg_steak_price:.2f}")
```

```python
#5. Determine which item type, chicken or steak, produced more total revenue (sum of item_price). 
# Total revenue for chicken
chicken_revenue = chicken_items['item_price'].sum()

# Total revenue for steak
steak_revenue = steak_items['item_price'].sum()

# Determine which produced more revenue
if chicken_revenue > steak_revenue:
    winner = "Chicken"
else:
    winner = "Steak"

print(f"\nTotal revenue from chicken items: ${chicken_revenue:,.2f}")
print(f"Total revenue from steak items: ${steak_revenue:,.2f}")
print(f"👉 {winner} items produced more revenue.")

```

```python
#6. Count the total number of missing values in the entire dataset. Also, display how many missing values exist in each individual column. 

# Count total missing values
total_missing = df.isnull().sum().sum()

# Count missing values by column
missing_by_column = df.isnull().sum()

print(f"\nTotal number of missing values in dataset: {total_missing}")
print("\nMissing values per column:")
print(missing_by_column)

```
