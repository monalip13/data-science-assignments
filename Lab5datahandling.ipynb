```python
#3.12 Run the following code to get player and team IDs: 
from nba_api.stats.static import players 
# Get all NBA players and find LeBron James ID
player_dict = players.get_players() 

bron = [player for player in player_dict if player['full_name'] == 'LeBron James'][0] 

bron_id = bron['id'] 
print("LeBron James ID:", bron_id)

# Get all NBA teams and find Golden State Warriors ID
from nba_api.stats.static import teams 

teams = teams.get_teams() 


GSW = [x for x in teams if x['full_name'] == 'Golden State Warriors'][0] 

GSW_id = GSW['id'] 
print("Golden State Warriors ID:", GSW_id)
```

```python
#3.13 Using what you’ve learned, collect game data for a player of your choice and export it to a CSV file. 

# Collect game data for Stephen Curry
curry = [player for player in player_dict if player['full_name'] == 'Stephen Curry'][0]
curry_id = curry['id']
gamelog_curry = playergamelog.PlayerGameLog(player_id=curry_id, season='2023')
gamelog_df = gamelog_curry.get_data_frames()[0]
print("Stephen Curry Game Data:")
print(gamelog_df.head())


# Export Stephen Curry's game logs to CSV
gamelog_df.to_csv("stephen_curry_games.csv", index=False)
```
