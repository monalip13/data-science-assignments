```python
# ==========================================
# RealEstate Linear Regression Visualization
# ==========================================

import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load dataset
df = pd.read_csv("/Users/patrickpatel/Downloads/Lab 11 - RealEstate.csv")

# Define variables
price_col = "Y house price of unit area"
model_a_pred = "X2 house age"
model_b_preds = ['X2 house age', 'X3 distance to the nearest MRT station', 'X4 number of convenience stores']

# ===============================
# Model A (Simple Linear Regression)
# ===============================
df_a = df[[price_col, model_a_pred]].dropna()
X_a = sm.add_constant(df_a[model_a_pred])
y_a = df_a[price_col]
model_a = sm.OLS(y_a, X_a).fit()
print("Model A summary:\n", model_a.summary())

# ===============================
# Model B (Multiple Linear Regression)
# ===============================
df_b = df[[price_col] + model_b_preds].dropna()
X_b = sm.add_constant(df_b[model_b_preds])
y_b = df_b[price_col]
model_b = sm.OLS(y_b, X_b).fit()
print("Model B summary:\n", model_b.summary())

# ===============================
# 1. Scatter Plot with Regression Line (Model A)
# ===============================
plt.figure(figsize=(8, 6))
sns.regplot(x=model_a_pred, y=price_col, data=df_a, line_kws={'color': 'red'})
plt.title("Model A: Relationship between House Age and Price per Unit Area")
plt.xlabel("House Age (Years)")
plt.ylabel("Price per Unit Area")
plt.grid(True)
plt.tight_layout()
plt.show()

# ===============================
# 2. Residual Plot (Model A)
# ===============================
plt.figure(figsize=(8, 6))
sns.residplot(x=model_a.predict(X_a), y=model_a.resid, lowess=True,
              line_kws={'color': 'orange'}, scatter_kws={'alpha': 0.6})
plt.title("Model A Residuals vs. Fitted Values")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.grid(True)
plt.tight_layout()
plt.show()

# ===============================
# 3. Actual vs Predicted Plot (Model B)
# ===============================
y_pred_b = model_b.predict(X_b)
plt.figure(figsize=(8, 6))
plt.scatter(y_b, y_pred_b, alpha=0.6, color='blue')
plt.plot([y_b.min(), y_b.max()], [y_b.min(), y_b.max()], 'r--', lw=2)
plt.title("Model B: Actual vs. Predicted House Prices")
plt.xlabel("Actual Price per Unit Area")
plt.ylabel("Predicted Price per Unit Area")
plt.grid(True)
plt.tight_layout()
plt.show()

# ===============================
# 4. Residuals Plot (Model B)
# ===============================
plt.figure(figsize=(8, 6))
sns.residplot(x=y_pred_b, y=model_b.resid, lowess=True,
              line_kws={'color': 'green'}, scatter_kws={'alpha': 0.6})
plt.title("Model B Residuals vs. Fitted Values")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.grid(True)
plt.tight_layout()
plt.show()

# ===============================
# 5. Q–Q Plot of Residuals (Model B)
# ===============================
sm.qqplot(model_b.resid, line='45', fit=True)
plt.title("Model B: Q–Q Plot of Residuals")
plt.tight_layout()
plt.show()

# ===============================
# 6. Coefficient Comparison Bar Chart (Model B)
# ===============================
coeffs = model_b.params[1:]  # exclude constant
plt.figure(figsize=(8, 6))
sns.barplot(x=coeffs.values, y=coeffs.index, palette="viridis")
plt.title("Model B Coefficient Magnitudes")
plt.xlabel("Coefficient Value")
plt.ylabel("Predictor Variable")
plt.grid(axis='x')
plt.tight_layout()
plt.show()

# ===============================
# 7. Correlation Heatmap (Optional)
# ===============================
plt.figure(figsize=(8, 6))
sns.heatmap(df_b.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of RealEstate Variables")
plt.tight_layout()
plt.show()
```
