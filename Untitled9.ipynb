```python
import pandas as pd

# Load your CSV
df = pd.read_csv("/Users/patrickpatel/Downloads/Lab 12 - Marketing.csv", sep=',')  # usually bank marketing uses ';'

# Inspect the first few rows and column names
print("Columns in CSV:\n", df.columns)
print("\nFirst 5 rows:\n", df.head())
print("\nValue counts for all columns:")
for col in df.columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts(dropna=False))
```

```python
# ---------------------------
# 3. Identify target column
# ---------------------------
# Manually set the target column after inspecting df.head() or df.columns
target_col = "y"  # Replace with the correct column name in your CSV
df['y_bin'] = df[target_col].map({'yes': 1, 'no': 0})

# Drop rows where target is missing
df = df.dropna(subset=['y_bin'])
```

```python
# ---------------------------
# 4. Define Features
# ---------------------------
numeric_cols = [
    'age', 'duration', 'campaign', 'pdays', 'previous',
    'emp_var_rate', 'cons_price_idx', 'cons_conf_idx', 'euribor3m', 'nr_employed'
]

cat_cols = [
    'job', 'marital', 'education', 'default', 'housing',
    'loan', 'contact', 'month', 'poutcome'
]

# Keep only valid columns present in the dataset
numeric_cols = [c for c in numeric_cols if c in df.columns]
cat_cols = [c for c in cat_cols if c in df.columns]

X = df[numeric_cols + cat_cols].copy()
y = df['y_bin']
```

```python
# ---------------------------
# 5. Handle Categorical Variables
# ---------------------------
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True)

# Fill any remaining missing values with 0
X_encoded = X_encoded.fillna(0)

print("Number of features after encoding:", X_encoded.shape[1])
```

```python
# ---------------------------
# 6. Train-Test Split
# ---------------------------
from sklearn.model_selection import train_test_split
RANDOM_STATE = 42

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.25, random_state=RANDOM_STATE, stratify=y
)

print("Train/Test shapes:", X_train.shape, X_test.shape, y_train.shape, y_test.shape)

```
