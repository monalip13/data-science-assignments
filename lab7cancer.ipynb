```python
# Lab 7: Cancer Data Analysis
# Author: Monaliben Patel
# Date: 2025-09-23
#import pandas library 
import pandas as pd

# Question 1: Read in the dataset Cancer.csv

# Read the CSV file saved from the original Excel file
df = pd.read_csv('Cancer.csv')
# Explanation: pd.read_csv() reads the CSV file into a pandas DataFrame called df
# Output: df contains all the data from the Cancer dataset
print("Data loaded successfully. First 5 rows:\n", df.head())

```

```python
# Question 2: Display the data types of each column

# Use the dtypes attribute to view the data type of each column
print("\nData types of each column:\n", df.dtypes)
# Explanation: Shows whether each column is integer, float, or object (string)
# This helps in understanding which columns may need conversion before data analysis

```

```python
# Question 3: Use the describe() function to generate summary statistics
summary_stats = df.describe(include='all')
print("\nSummary statistics of the dataset:\n", summary_stats)
# Explanation: describe() provides count, mean, std, min, max, and quartiles for numeric columns
# For object columns, it shows unique values, top value, and frequency

```

```python
# Question 4: Group the dataset by 'class' and 'doctor_name'

grouped = df.groupby(['class', 'doctor_name'])
# Display the size of each group
print("\nGrouped data by class and doctor_name counts:\n", grouped.size())
# Explanation: groupby() organizes data by class and doctor_name, size() counts rows in each group
# Output: Number of patients per doctor per class is displayed

```

```python
# Question 5: Provide a summary of missing values within the dataset
missing_summary = df.isnull().sum()
print("\nMissing values per column:\n", missing_summary)
# Explanation: isnull() checks for missing data, sum() counts them per column
# Output: Number of missing values in each column is displayed

# Expand summary of missing values
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100
total_missing = missing_count.sum()
columns_with_missing = missing_count[missing_count > 0].index.tolist()

print("Missing Values Summary:")
print("------------------------")
print("Number of missing values per column:\n", missing_count)
print("\nPercentage of missing values per column:\n", missing_percent.round(2))
print("\nTotal missing values in the dataset:", total_missing)
print("\nColumns with missing values:", columns_with_missing)
```

```python
# Question 6: Drop all rows that contain missing data
df_clean = df.dropna()
print("\nData shape after dropping rows with missing values:", df_clean.shape)
# Explanation: dropna() removes any row with at least one missing value
# Output: Shows new dataset shape after cleaning

```

```python
# Question 7: Summarize the number of unique values in each column
unique_values = df_clean.nunique()
print("\nNumber of unique values per column:\n", unique_values)
# Explanation: nunique() counts unique entries per column
# Output: Can highlight columns with repeated values like patient_id or doctor_name

# Interesting Findings
#An interesting finding from the unique value summary is that patient_id has fewer unique values than the total number of rows, which indicates that some patients appear multiple times in the dataset. 
#In contrast, doctor_name has only a handful of unique values, suggesting that only a few doctors are responsible for diagnosing many patients. 
#Additionally, the class column has exactly two unique values, confirming that this dataset represents a binary classification problem (likely benign vs. malignant). 
#For numeric attributes such as cell_size_uniformity and cell_shape_uniformity, the number of unique values is relatively small, which suggests they are ordinal categorical ratings (e.g., scores from 1 to 10) rather than continuous measurements.
```

```python
# Question 8: Find duplicate values in 'patient_id'. Which patient_id appears most frequently?
duplicates = df_clean['patient_id'].value_counts()
most_frequent_patient = duplicates.idxmax()
print("\nMost frequent patient_id:", most_frequent_patient, "appears", duplicates.max(), "times")
# Explanation: value_counts() counts each patient_id occurrence, idxmax() finds the most frequent
# Output: Identifies patient with most repeated records

```

```python
# Question 9: Remove patients (rows) where the 'patient_id' appears more than two times
df_filtered = df_clean[df_clean['patient_id'].map(df_clean['patient_id'].value_counts()) <= 2]
print("\nData shape after filtering patient_id > 2 occurrences:", df_filtered.shape)
# Explanation: map(value_counts()) replaces patient_id with its occurrence count
# Filtering keeps only rows where patient_id occurs 1 or 2 times
# Output: Dataset shape reduced to remove overrepresented patients
df_filtered


```

```python
# Question 10: Create a new DataFrame called 'categorical_df' containing 'patient_id' and 'doctor_name'.
# Add a column 'doctor_count' with value 1
categorical_df = df_filtered[['patient_id', 'doctor_name']].copy()
categorical_df['doctor_count'] = 1
print("\nCategorical DataFrame with doctor_count:\n", categorical_df.head())
# Explanation: Isolate patient_id and doctor_name for one-hot encoding
# doctor_count=1 will be used to count visits when pivoted

```

```python
# Question 11: Pivot the DataFrame so that each doctor's name becomes a column and the cell values indicate visits
# Fill missing cells with 0
pivot_df = categorical_df.pivot_table(index='patient_id', columns='doctor_name',
                                      values='doctor_count', aggfunc='sum', fill_value=0)
print("\nPivot table (one-hot encoding of doctors):\n", pivot_df.head())
# Explanation: pivot_table() reshapes long-to-wide format; each doctor becomes a column
# fill_value=0 ensures missing doctor visits are recorded as 0
# Output: Each row = patient, each column = doctor, values = number of visits

```

```python
# Question 12: Drop the multi-index from columns using droplevel()
# Check if columns are MultiIndex
if isinstance(pivot_df.columns, pd.MultiIndex):
    pivot_df.columns = pivot_df.columns.droplevel(0)
print("\nPivot table after removing multi-index:\n", pivot_df.head())
# Explanation: Sometimes pivot_table creates MultiIndex; droplevel flattens it

```

```python
# Question 13: Display the one-hot encoded DataFrame
print("\nOne-hot encoded DataFrame:\n", pivot_df.head())
# Explanation: Displays transformed data ready for merging

```

```python
# Question 14: Join this one-hot encoded DataFrame back to the original dataset using merge()
df_combined = df_filtered.merge(pivot_df, on='patient_id', how='left')
print("\nCombined DataFrame after merge:\n", df_combined.head())
# Explanation: merge() joins one-hot encoded doctor info with original data on patient_id
# Output: Original columns + one-hot doctor columns

```

```python
# Question 15: Display the final combined DataFrame
print("\nFinal combined DataFrame:\n", df_combined.head())
# Explanation: Shows merged dataset

```

```python
# Question 16: Drop the 'doctor_name' column from the combined DataFrame
df_combined.drop(columns=['doctor_name'], inplace=True)
print("\nFinal DataFrame after dropping 'doctor_name':\n", df_combined.head())
# Explanation: doctor_name is now redundant after one-hot encoding

```

```python

# Question 17: Define a function named celltypelabel(x) to classify a patient’s cell as 'normal' or 'abnormal'

def celltypelabel(x):
    """
    Classify patient's cell as 'normal' or 'abnormal'
    based on 'cell_size_uniformity' and 'cell_shape_uniformity'.
    """
    if (x['cell_size_uniformity'] > 5) & (x['cell_shape_uniformity'] > 5):
        return 'normal'
    else:
        return 'abnormal'

```

```python
# Question 18: Use the apply() function to create a new column 'cell_type_label'

df_combined['cell_type_label'] = df_combined.apply(celltypelabel, axis=1)
print("\nDataFrame with 'cell_type_label' column:\n", df_combined.head())
# Explanation: apply() applies function row-wise (axis=1)
# Output: Each patient now has a cell_type_label: 'normal' or 'abnormal'


# Save final cleaned and transformed dataset

df_combined.to_csv('Cancer_Cleaned_Combined.csv', index=False)
print("\nFinal cleaned and transformed dataset saved as 'Cancer_Cleaned_Combined.csv'")
# Explanation: Saves dataset for reporting or further analysis

# Additional Summary: Count of normal vs abnormal classifications
classification_summary = df_combined['cell_type_label'].value_counts()

print("\nSummary of Cell Type Classification:")
print(classification_summary)
#Explanation 
#The new column 'cell_type_label' now marks each patient’s cell type as either "normal" or "abnormal".

#The summary table shows how many patients fall into each group.

#In this example output, most patients were classified as abnormal (555), while 121 were classified as normal.
```
