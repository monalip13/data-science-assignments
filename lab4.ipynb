```python
import pandas as pd
import numpy as np
```

```python
# Q1: Create and display a one-dimensional array-like object (Series) with any numerical or string data of your choice. 
custom_series = pd.Series([5, 10, 15, 20, 25])       # Creating a Pandas Series with numeric data
print("Q1: Custom Series:\n", custom_series)         # Displaying the Series
print("\n# Explanation: Created a Pandas Series with numeric data.\n")
```

```python
# Q2: Create two Pandas Series: [2, 4, 6, 8, 10] and [1, 3, 5, 7, 9]. Perform addition, subtraction, multiplication, and division on them and display the results. 
series1 = pd.Series([2, 4, 6, 8, 10])       # First numerical Series
series2 = pd.Series([1, 3, 5, 7, 9])        # Second numerical Series
print("Series 1:\n", series1)               # Display Series1
print("Series 2:\n", series2)               # Display Series2
```

```python
# Addition
print("\nAddition:\n", series1 + series2)       # Element-wise addition
```

```python
# Subtraction
print("\nSubtraction:\n", series1 - series2)      # Element-wise subtraction
```

```python
# Multiplication
print("\nMultiplication:\n", series1 * series2)      
```

```python
# Division
print("\nDivision:\n", series1 / series2)            # Element-wise division
print("\n# Explanation: Element-wise arithmetic operations performed correctly.\n")
```

```python

df = pd.read_excel('Automobile.xls', engine='xlrd')   # Reading the Excel file into a DataFrame
```

```python
df
```

```python
# Q4: Print the first five and last five rows to understand the structure. 
print("First 5 rows:\n", df.head())        # Display first 5 rows
print("\nLast 5 rows:\n", df.tail())       # Display last 5 rows    
print("\n# Explanation: Displayed top and bottom rows to inspect dataset.\n")
```

```python
# Q5: Remove any records containing '?', 'n.a', or NaN values. Save the cleaned DataFrame to a new CSV file. 
df_cleaned = df.replace('?', np.nan).replace('n.a', np.nan).dropna()   # Replace unknown values with NaN and drop rows with NaN
df_cleaned.to_csv('Automobile_Cleaned.csv', index=False)               # Save cleaned DataFrame to CSV
print("Cleaned DataFrame saved as 'Automobile_Cleaned.csv'\n")
print("# Explanation: Removed missing/unknown values and saved cleaned dataset.\n")
```

```python
# Q6: Find and print the most expensive car in the dataset. Show both the price and the company name. 
df_cleaned['price'] = df_cleaned['price'].astype(float)             # Convert price column to float
most_expensive = df_cleaned.loc[df_cleaned['price'].idxmax()]       # Get row with maximum price  
print("Most expensive car:")
print("Company:", most_expensive['company'], "Price:", most_expensive['price'])
print("\n# Explanation: Identified the car with highest price.\n")
```

```python
# Q7: Print all the details of cars manufactured by Toyota. 
toyota_cars = df_cleaned[df_cleaned['company'].str.lower() == 'toyota']          # Filter rows where company is Toyota
print("Toyota Cars Details:\n", toyota_cars)
print("\n# Explanation: Filtered rows where company is Toyota.\n")
```

```python
# Q8: Count and display the total number of cars per company. 
car_counts = df_cleaned['company'].value_counts()                       # Count number of cars per company
print("Total number of cars per company:\n", car_counts)
print("\n# Explanation: Counted cars for each manufacturer.\n")
```

```python
# Q9:  Display the most expensive car from each company. 
most_exp_by_company = df_cleaned.loc[df_cleaned.groupby('company')['price'].idxmax()]           # Group by company and get max price row
print("Most expensive car from each company:\n", most_exp_by_company)
print("\n# Explanation: Grouped by company and selected max price.\n")
```

```python
# Q10:  Calculate the average mileage (fuel efficiency) for each car company. 
df_cleaned['average-mileage'] = df_cleaned['average-mileage'].astype(float)              # Convert mileage to float
avg_mileage = df_cleaned.groupby('company')['average-mileage'].mean()                    # Calculate average mileage per company
print("Average mileage per company:\n", avg_mileage)                                     # Display average mileage
print("\n# Explanation: Calculated mean fuel efficiency per manufacturer.\n")
```

```python
# Q11: Sort all cars in ascending order based on the price column and display the result. 
sorted_cars = df_cleaned.sort_values(by='price')                     # Sort DataFrame by price in ascending order
print("Cars sorted by price (ascending):\n", sorted_cars)            # Display sorted DataFrame
print("\n# Explanation: Sorted DataFrame based on price column.\n")
```

```python
# Q12:  Use the provided dictionaries to create two DataFrames and concatenate them. 

GermanCars = {'Company': ['Ford', 'Mercedes', 'BMV', 'Audi'], 
              'Price': [23845, 171995, 135925, 71400]}
japaneseCars = {'Company': ['Toyota', 'Honda', 'Nissan', 'Mitsubishi'], 
                'Price': [29995, 23600, 61500, 58900]}

df_german = pd.DataFrame(GermanCars)           # Create DataFrame from GermanCars
df_japanese = pd.DataFrame(japaneseCars)       # Create DataFrame from japaneseCars

concatenated_cars = pd.concat([df_german, df_japanese], ignore_index=True)             # Concatenate DataFrames vertically
print("Concatenated DataFrame:\n", concatenated_cars)                                  # Display concatenated DataFrame
print("\n# Explanation: Combined German and Japanese cars vertically.\n")
```

```python
# Q13:  Create and merge the Car_Price and Car_Horsepower DataFrames on the Company column and display the result. 
Car_Price = {'Company': ['Toyota', 'Honda', 'BMV', 'Audi'], 
             'Price': [23845, 17995, 135925, 71400]}
Car_Horsepower = {'Company': ['Toyota', 'Honda', 'BMV', 'Audi'], 
                  'horsepower': [141, 80, 182, 160]}

df_price = pd.DataFrame(Car_Price)             # DataFrame for car prices
df_hp = pd.DataFrame(Car_Horsepower)           # DataFrame for car horsepower

merged_df = pd.merge(df_price, df_hp, on='Company')          # Merge on 'Company' column
print("Merged DataFrame (Price + Horsepower):\n", merged_df)   # Display merged DataFrame
print("\n# Explanation: Merged DataFrames on 'Company' to show price and horsepower.\n")
```
