```python
# Lab 8 - Treatment Analysis
# Student: Monaliben Patel
# Date: 2025-09-25

# Question 1: Read the dataset 'Treatment.csv' into a DataFrame using Pandas

import pandas as pd
df = pd.read_csv('Treatment.csv')

# Display the first few rows to confirm the data loaded correctly
print("First 5 rows of the dataset:")
print(df.head())

# Explanation:
# pd.read_csv() reads the CSV file into a Pandas DataFrame
# df.head() shows the first 5 rows to verify the dataset


```

```python
print("Columns in the dataset:")
print(df.columns.tolist())
```

```python
 
# Question 2: Group the dataset by the 'Treatment' column

grouped_treatment = df.groupby('Treatment')
grouped_treatment
# Explanation:
# groupby('Treatment') creates groups based on unique values in the 'Treatment' column

```

```python
# Question 3: Analyze the distribution of 'Relative Fitness' within each treatment group

treatment_summary = grouped_treatment['RelativeFitness'].describe()
print("\nDistribution of 'Relative Fitness' by Treatment:")
print(treatment_summary)

# Explanation:
# describe() computes count, mean, std, min, 25%, 50%, 75%, and max
# This allows observation of spread, central tendency, and variability in each treatment group

# Observations:
# Some treatments may have higher average relative fitness than others.
# Standard deviation shows variability within each treatment.
# Min and max indicate the range of fitness values for each group.

```

```python
# Question 4: Group the dataset by both 'Treatment' and 'Group'

multi_grouped = df.groupby(['Treatment', 'Group'])
multi_summary = multi_grouped['RelativeFitness'].describe()
print("\nDistribution of 'Relative Fitness' by Treatment and Group:")
print(multi_summary)

# Explanation:
# Multi-level grouping allows analysis of subgroups within each treatment
# This shows if different groups respond differently within the same treatment

# Observations:
# Compare means and standard deviations across groups
# Check if the relative fitness differences are consistent across groups
# Identify any unusual patterns or outliers in specific groups

```

```python
# Question 5: Apply aggregation functions: mean, max, median, sum, std

aggregated = multi_grouped['RelativeFitness'].agg(['mean', 'max', 'median', 'sum', 'std'])
print("\nAggregated statistics for 'Relative Fitness' by Treatment and Group:")
print(aggregated)

# Explanation:
# agg() allows multiple aggregation functions to be applied at once
# mean, max, median, sum, std provide a comprehensive statistical summary
# Useful for comparing central tendency, spread, and total fitness across groups

# Observations:
# Mean and median indicate central tendency; differences may show skewed distributions
# Max shows the peak fitness achieved
# Sum gives the total fitness within each subgroup
# Standard deviation highlights variability within subgroups
# Potential Outliers
#By comparing max values to the upper quartile in the describe() output, we can see which treatments had unusually high performers. 
#These may be worth investigating further, as they could represent special conditions, errors, or hidden advantages.
```
