```python
# 2. Load the dataset into a Pandas DataFrame. Display the first 10 rows and a screenshot of the dataset structure. 
# Read CSV into pandas
import io
import pandas as pd  # for data manipulation
import matplotlib.pyplot as plt  # for plotting
import seaborn as sns  # for advanced plotting
df = pd.read_csv("/Users/patrickpatel/Downloads/TSLA.csv")

# Display head(10)
display(df.head(10))

# Show data structure
buf = io.StringIO()
df.info(verbose=True, buf=buf)
print(buf.getvalue())

# Explanation 
# pd.read_csv(...) — loads the CSV into a DataFrame.
# display(df.head(10)) — shows the first 10 rows.
# df.info(...) captured to a string buffer so we can see column names, non-null counts, and dtypes
# The CSV had 2,956 rows and 7 columns (for example: Date, Open, High, Low, Close, Adj Close, Volume 
# — depending on the file; see the printed df.info()
```

```python
# 3. Convert the date/time column into a proper datetime format using Pandas. 
# Identify date-like columns and convert to datetime

# Display first 10 rows
print("First 10 rows of dataset:")
display(df.head(10))

# Check column types to confirm 'Date' is not datetime
print("\nData types before conversion:")
print(df.dtypes)

# Convert the 'Date' column to datetime type
df['Date'] = pd.to_datetime(df['Date'])

# Verify conversion
df.info()

# Explanation:
# pd.to_datetime() ensures the 'Date' column is recognized as a datetime object.
# This is crucial for time series indexing, plotting, and resampling.
```

```python
# 4. Set the datetime column as the index of your DataFrame. 
# Set 'Date' column as the index of the DataFrame
df.set_index('Date', inplace=True)

# Display first 5 rows to confirm index
df.head()
# Explanation:
# set_index('Date', inplace=True) makes the Date column the index, which is required for time series operations.
```

```python
# 5. Generate and plot a time series line graph using matplotlib or seaborn. 
#Show trends over time and explain them with help from the AI. 

import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_style('whitegrid')

# Plot Tesla 'Close' price over time
plt.figure(figsize=(12,6))
plt.plot(df.index, df['Close'], color='blue', label='Close Price')
plt.title('Tesla Stock Closing Price Over Time')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()

# Explanation:
# plt.plot() plots the closing price of Tesla over time.
# Tesla stock shows a strong long-term upward trend from 2010 to 2021.
# Several short-term fluctuations and sharp peaks/dips are evident, reflecting high market volatility.
# Notable price spikes in 2020–2021 correspond to product announcements, stock splits, and investor optimism.
# The AI helped interpret these fluctuations as events influencing stock trends rather than random noise.
```

```python
# 6. Use rolling averages (such as 7-day or 30-day windows) to smooth the data. Plot and interpret the results. 
# Calculate 7-day rolling average
df['Rolling_7d'] = df['Close'].rolling(window=7).mean()

# Calculate 30-day rolling average
df['Rolling_30d'] = df['Close'].rolling(window=30).mean()

# Plot original and rolling averages
plt.figure(figsize=(12,6))
plt.plot(df.index, df['Close'], label='Original Close', alpha=0.5)
plt.plot(df.index, df['Rolling_7d'], label='7-Day Rolling Avg', color='orange')
plt.plot(df.index, df['Rolling_30d'], label='30-Day Rolling Avg', color='red')
plt.title('Tesla Stock Closing Price with Rolling Averages')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()


# Explanation:
# rolling(window) smooths short-term fluctuations.
# 7-day average captures weekly trends; 30-day captures monthly trends.
# Insight: Rolling averages highlight the long-term upward trend and reduce noise, revealing true patterns.
```

```python
# 7. Resample the data (by week or month) and visualize changes in trends. 
# Ask your AI tool what transformations might be useful. 
# Resample Tesla stock data monthly (mean closing price)
# Resample Tesla stock data monthly using 'ME' (Month End)
df_monthly = df['Close'].resample('ME').mean()

# Plot monthly average closing price
plt.figure(figsize=(12,6))
plt.plot(df_monthly.index, df_monthly, marker='o', color='green')
plt.title('Monthly Average Closing Price of Tesla Stock')
plt.xlabel('Month')
plt.ylabel('Average Close Price (USD)')
plt.grid(True)
plt.show()


# Explanation:
# 'ME' ensures resampling aligns with the end of each month.
# 'MS' could be used if you want the resampling aligned with start of the month.
# The rest of the analysis remains exactly the same; only the resampling string changes to avoid future warnings.
```

```python
# 8. Check for any seasonal patterns, outliers, or anomalies using visual inspection and AI insights. 

# Visual inspection for outliers
plt.figure(figsize=(12,6))
sns.boxplot(x=df.index.year, y=df['Close'])
plt.title('Tesla Stock Closing Price Distribution by Year')
plt.xlabel('Year')
plt.ylabel('Close Price (USD)')
plt.show()

# Explanation:
# Boxplot identifies yearly trends, volatility, and outliers.
# AI Insight: Major anomalies in 2020-2021 reflect pandemic effects, stock splits, and news events.
# Seasonal patterns: Some recurring high activity occurs around product announcements or market cycles.
```
